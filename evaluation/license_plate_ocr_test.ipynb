```python
!pip install ultralytics -q
from ultralytics import YOLO

# 1. Khởi tạo mô hình với đường dẫn file best.pt bạn lấy từ mục Input
path_to_best_pt = "/kaggle/input/notebooks/toquanghan/license-plate-ocr/Ket_Qua_OCR/license_plate_ocr_yolo26m/weights/best.pt"
model = YOLO(path_to_best_pt)

print("🚀 Bắt đầu chấm điểm OCR trên tập Test...")

# 2. Chạy đánh giá (Validation) với chỉ định ép buộc dùng tập 'test'
metrics = model.val(
    data="/kaggle/input/notebooks/toquanghan/license-plate-ocr/License-Plate---OCR-3/data.yaml",
    split="test",
    project="/kaggle/working/Ket_Qua_Test_OCR",
    name="danh_gia",
    exist_ok=True
)

print("✅ Đánh giá hoàn tất! Kết quả chi tiết đã được lưu trong thư mục Ket_Qua_Test_OCR.")
```
