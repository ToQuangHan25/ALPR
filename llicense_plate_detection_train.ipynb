```python
# 1. Cài đặt thư viện (Chỉ cần chạy 1 lần)
!pip install ultralytics -q

from ultralytics import YOLO
import os

# 2. Khởi tạo mô hình bản Medium
model = YOLO("yolo26m.pt")

# 3. Cấu hình huấn luyện tối ưu
# - batch=64: Tận dụng tối đa 30GB VRAM (2x15GB) của T4x2
# - device="0,1": Kích hoạt huấn luyện song song (DDP) trên cả 2 card
# - resume=True: Tự động tiếp tục nếu bị ngắt kết nối giữa chừng
# - imgsz=640: Độ phân giải tiêu chuẩn, cân bằng giữa chi tiết và tốc độ
print("🚀 Đang khởi động tiến trình huấn luyện chuyên sâu...")

results = model.train(
    data="/kaggle/input/datasets/toquanghan/license-plate/dataset_yolo/data.yaml",
    epochs=100,
    patience=15,
    imgsz=640,
    batch=32,
    device="0,1",
    workers=4,
    project="/kaggle/working/Ket_Qua_Train",
    name="license_plate_detection_yolo26m",
    exist_ok=True,
    resume=True,

    # --- Fine-tuning để tăng độ chính xác ---
    lr0=0.001,
    lrf=0.01,
    close_mosaic=10,
    augment=True
)

print("✅ Huấn luyện hoàn tất! File trọng số đã được lưu trong thư mục Ket_Qua_Train.")
```
