```python
# 1. Cài đặt thư viện
!pip install ultralytics roboflow -q

# 2. Tải dữ liệu từ Roboflow
from roboflow import Roboflow
from ultralytics import YOLO
import os

rf = Roboflow()
project = rf.workspace("license-plate-vn").project("license-plate-ocr-hugcj")
version = project.version(3)
dataset = version.download("yolo26")

model = YOLO("yolo26m.pt")

print("🚀 Đang khởi động tiến trình huấn luyện OCR chuyên sâu...")
results = model.train(
    data=f"{dataset.location}/data.yaml",
    epochs=100,
    patience=15,
    imgsz=640,
    batch=32,
    device="0,1",
    workers=4,
    project="/kaggle/working/Ket_Qua_OCR",
    name="license_plate_ocr_yolo26m",
    exist_ok=True,
    resume=True,

    # --- Fine-tuning để tăng độ chính xác ---
    lr0=0.001,
    lrf=0.01,
    close_mosaic=10,
    augment=True
)

print("✅ Huấn luyện OCR hoàn tất! File trọng số đã được lưu trong thư mục Ket_Qua_OCR.")
```
